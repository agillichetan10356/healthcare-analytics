## Create Patient Dataset

```python
patient_data = [
    (101, "John", 45, "Diabetes"),
    (102, "Mary", 32, "Asthma"),
    (103, "David", 65, "Hypertension"),
    (104, "Sarah", 29, "Asthma"),
    (105, "Robert", 72, "Heart Disease"),
    (106, "Emma", 55, "Diabetes")
]

columns = [
    "patient_id",
    "patient_name",
    "age",
    "diagnosis"
]

df = spark.createDataFrame(patient_data, columns)

display(df)

```

## Create Bronze Layer

```python
# Save Bronze Table

df.write.mode("overwrite").saveAsTable("bronze_patients")
```
