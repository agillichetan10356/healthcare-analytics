## Create Silver Layer

```python
# Read Bronze Table

df = spark.table("bronze_patients")

display(df)
```

```python
# Remove Nulls

df_clean = df.dropna()
```

```python
# Remove Duplicates

df_clean = df_clean.dropDuplicates()
```

```python
# Save Silver Table

df_clean.write.mode("overwrite").saveAsTable("silver_patients")
```

```python
# Verify Silver Table

display(
    spark.table("silver_patients")
)
```
