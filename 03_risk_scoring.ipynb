## Create Gold Layer

```python
# Read Silver Table

from pyspark.sql.functions import when,col

df = spark.table("silver_patients")
```

```python
# Create Risk Categories

df_risk = (
    df.withColumn(
        "risk_category",
        when(col("age") > 60, "HIGH")
        .when(col("age") >= 40, "MEDIUM")
        .otherwise("LOW")
    )
)

display(df_risk)
```

```python
# Save Gold Table

df_risk.write.mode("overwrite").saveAsTable("gold_patient_risk")
```

```python
# Verify Gold Table

display(
    spark.table("gold_patient_risk")
)
```
